<h3>Step_1_3: Identify primary and foreign keys</h3>

```python
# define a dictionary of dataset
datasets = {'category_translation':category_translation,
            'customers':customers,
            'location':location,
            'order_item':order_item,
            'orders':orders,
            'payments':payments,
            'products':products,
            'reviews':reviews,
            'sellers':sellers
           }

# function to check possible primary keys
def get_primary_keys(ds):
    possible_keys = []

    for col in ds.columns:
        # Check unique + no null values
        if ds[col].is_unique and ds[col].isnull().sum() == 0:
            possible_keys.append(col)

    return possible_keys


# Check all datasets
for name, ds in datasets.items():
    keys = get_primary_keys(ds)

    print(f"\nDataset: {name}")

    if keys:
        print("Possible Primary Key(s):", keys)
    else:
        print("No single-column primary key found.")


```

<h3>Step_1_4: Understand relationships between tables</h3>

```python
# Store relationship results
relationships = []

# Compare each dataset with every other dataset
for name1, df1 in datasets.items():
    for name2, df2 in datasets.items():

        if name1 != name2:
            # Find common columns
            common_cols = set(df1.columns).intersection(set(df2.columns))

            for col in common_cols:

                # Unique checks
                unique1 = df1[col].is_unique
                unique2 = df2[col].is_unique

                # Decide relationship type
                if unique1 and unique2:
                    relation = "One-to-One"
                elif unique1 and not unique2:
                    relation = "One-to-Many"
                elif not unique1 and unique2:
                    relation = "Many-to-One"
                else:
                    relation = "Many-to-Many"

                # Save result
                relationships.append({
                    "Table_1": name1,
                    "Table_2": name2,
                    "Common_Column": col,
                    "Relationship": relation
                })

# Convert to DataFrame
relationship_df = pd.DataFrame(relationships)

# Remove duplicate reverse rows (optional)
relationship_df = relationship_df.drop_duplicates()

# Show output
print(relationship_df)
```

```python
orders = pd.read_csv("orders.csv")
# converting str datatype date column into datetime column
orders["order_approved_at"]=pd.to_datetime(orders["order_approved_at"])
orders["order_purchase_timestamp"]=pd.to_datetime(orders["order_purchase_timestamp"])
orders["order_delivered_carrier_date"]=pd.to_datetime(orders["order_delivered_carrier_date"])
orders["order_delivered_customer_date"]=pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_estimated_delivery_date"]=pd.to_datetime(orders["order_estimated_delivery_date"])
#orders.info()

####### impute order_approved_at
blank_approved_at= orders[(orders["order_status"]=="delivered") & (orders["order_approved_at"].isna())].index
# print(mask)
orders.loc[blank_approved_at, "order_approved_at"] = orders.loc[blank_approved_at, "order_purchase_timestamp"].values
#orders.isnull().sum()

####### impute order_delivered_carrier_date column 
# with the median of time diff between 
# filled order_approved_at and order_delivered_carrier_date 
# and order_status as delivered
# first get the rows with filled above two columns
filled_order_delivered =  orders[orders["order_approved_at"].notna() & orders["order_delivered_carrier_date"].notna() & (orders["order_status"] == "delivered")]

# calculate the median from approval to carrier
filled_median_sec = (filled_order_delivered["order_delivered_carrier_date"] - filled_order_delivered["order_approved_at"]).dt.total_seconds().median()
#print(filled_median_sec)
# get the blank order_delivered_carrier_date rows to be imputed with the median
blank_carrier_delivered =  orders[orders["order_approved_at"].notna() & orders["order_delivered_carrier_date"].isna() & (orders["order_status"] == "delivered")].index
#print(blank_carrier_delivered)
# impute blank order_delivered_carrier_date
orders.loc[blank_carrier_delivered, "order_delivered_carrier_date"] = (orders.loc[blank_carrier_delivered, "order_approved_at"] + pd.to_timedelta(filled_median_sec, unit="s")).values
#orders.isnull().sum()

###### impute order_delivered_customer_date column
# with the median date diff in days bwtween
# order_delivered_carrier_date and order_delivered_customer_date
# and order_status with delivered
# first get the rows with filled above two columns
filled_order_delivered =  orders[orders["order_delivered_carrier_date"].notna() & orders["order_delivered_customer_date"].notna() & (orders["order_status"] == "delivered")]
#filled_order_delivered.isnull().sum()
# calculate the median from carrier date to customer date
filled_median_days = (filled_order_delivered["order_delivered_customer_date"] - filled_order_delivered["order_delivered_carrier_date"]).dt.total_seconds().median()/86400
#print(filled_median_days)
# get the blank order_delivered_carrier_date rows to be imputed with the median
blank_carrier_delivered =  orders[orders["order_approved_at"].notna() & orders["order_delivered_carrier_date"].isna() & (orders["order_status"] == "delivered")].index
#print(blank_carrier_delivered)
# impute blank order_delivered_customer_date
orders.loc[blank_carrier_delivered, "order_delivered_customer_date"] = (orders.loc[blank_carrier_delivered, "order_delivered_carrier_date"] + pd.to_timedelta(filled_median_days, unit="D")).values
#orders.isnull().sum()
```

<h3>Step_3_1: Construct a Master Dataset by merging multiple tables

```python
# first create a copy of orders dataset as master dataset
master_df = orders.copy()

# 1.	orders + customers - merge customers
master_df = master_df.merge(
    customers,
    on="customer_id",
    how="left"
)


# 2.	orders + order_items  - merge order_items
master_df = master_df.merge(
    order_item,
    on="order_id",
    how="left"
)


# 3.	order_items + products   - merge products
master_df = master_df.merge(
    products,
    on="product_id",
    how="left"
)

# 4.	orders + payments    - merge payments
master_df = master_df.merge(
    payments,
    on="order_id",
    how="left"
)

# 5.	orders + reviews    - merge reviews
master_df = master_df.merge(
    reviews,
    on="order_id",
    how="left"
)
print(master_df.shape)
print(master_df.head())

# second merge order_item dataset with sellers dataset
# 6.	order_items + sellers    - merge 
order_item = order_item.merge(
    sellers,
    on="seller_id",
    how="left"
)
print(order_item.shape)
print(order_item.head())

# third merge products dataset with category_translation dataset
# 7.	products + category_translation   - merge 
products = products.merge(
    category_translation,
    on="product_category_name",
    how="left"
)
print(products.shape)
print(products.head())
```

<h1>Step 4: Feature Engineering</h1>

<h3>Step_4_1: Total order value (aggregated from order_items or payments) </h3>

```python
# A feature to have total order value
# first get the aggregate payment value
total_order_value = payments.groupby("order_id")["payment_value"].sum().reset_index()
# rename the payment_value column to total_order_value
total_order_value.rename(columns={"payment_value": "total_order_value"},inplace=True)
total_order_value.head()

# secondly merge the above table to orders table
orders = orders.merge(
    total_order_value,
    on="order_id",
    how="left"
)
orders.head()
```

<h3>Step_4_2: Delivery time (order purchase to delivery date)</h3>

```python
# To add a new column to orders table as delivery_time_days
orders["delivery_time_days"] = (orders["order_delivered_customer_date"]-orders["order_purchase_timestamp"]).dt.days
orders.head()
```

<h1>Step 4: Feature Engineering</h1>

<h3>Step_4_3: Number of items per order </h3>

```python
# first to create an object with the no. of items
items_per_order = order_item.groupby("order_id")["order_item_id"].count().reset_index()

# renaming the column
items_per_order.rename(
    columns={
        "order_item_id":
        "order_item_nos"
    },
    inplace=True
)
print(items_per_order)

# merge the above to orders dataset
orders = orders.merge(items_per_order,on="order_id",how="left")
orders.head()
```

<h1>Step 4: Feature Engineering</h1>

<h3>Step_4_4: Customer purchase frequency </h3>

```python
# add ne feature of customer purchase frequency
customer_purchase_frequency = orders.groupby("customer_id")["order_id"].count().reset_index()

# rename the column
customer_purchase_frequency.rename(
    columns={
        "order_id":
        "customer_purchase_frequency"
    },
    inplace=True
)

# merge with orders dataset
orders = orders.merge(
    customer_purchase_frequency,
    on="customer_id",
    how="left"
)
orders.head()
```

<h3>Step_5_1: Customer Analysis </h3>

```python
# for customer analysis of new and repeat customers
# we need to merge customer, orders and payments datasets
# first need to know the aggregate total payments per order
order_value = payments.groupby(
    "order_id"
)["payment_value"].sum().reset_index()
# orders.drop(columns=["payment_value_x"],inplace=True)
# orders.drop(columns=["payment_value_y"],inplace=True)
# orders.drop(columns=["payment_value"],inplace=True)

# merge with orders
orders = orders.merge(
    order_value,
    on="order_id",
    how="left"
)

orders.head()
# merge to master dataset
master_df = orders.merge(
    customers,
    on="customer_id",
    how="left"
)
master_df.head()
```

```python
# ------------------------------------- #
# ------ New vs Repeat Customers ------ #
# ------------------------------------- #
# Need to count the orders of each customer
# For customer with 1 order is a New Customer
# or more than 1 order is a Repeat Customer
customer_orders = master_df.groupby("customer_unique_id")["order_id"].count().reset_index()

customer_orders.rename(columns={"order_id": "total_orders"},inplace=True)

customer_orders["customer_type"] = (customer_orders["total_orders"].apply(
    lambda x:

        "Repeat Customer"

        if x > 1

        else "New Customer"
    )
)

customer_type_count = customer_orders["customer_type"].value_counts()

print(customer_type_count)

# graphical representation 
plt.figure(figsize=(6,4))

sns.countplot(data=customer_orders,x="customer_type")

plt.title("New vs Repeat Customers")
plt.xlabel("Type of Customer")
plt.ylabel("Customer Counts")
plt.show()
```

```python
# ----------------------------------------------- #
# ------ High-value vs Low-value Customers ------ #
# ----------------------------------------------- #

# Need to classify customers on their spending
# Calculate customer spending
customer_spending = master_df.groupby("customer_unique_id")["payment_value"].sum().reset_index()

# rename column
customer_spending.rename(columns={"payment_value":"customer_lifetime_value"},inplace=True)

# set spending threshold
threshold = customer_spending["customer_lifetime_value"].median()
print(threshold)

# classicfication of high and low value by threshold
customer_spending["customer_segment"] = (customer_spending["customer_lifetime_value"].apply(

        lambda x:
        "High Value"
    
        if x >= threshold

        else "Low Value"
    )
)

print(customer_spending["customer_segment"].value_counts())

# graphical representation
plt.figure(figsize=(6,4))

sns.countplot(data=customer_spending,x="customer_segment")
plt.title("High vs Low Value Customers")
plt.xlabel("Type of Customer")
plt.ylabel("Customer Spendings")
plt.show()
```

```python
# -------------------------------------------------- #
# ------ Geographic Distribution of Customers ------ #
# -------------------------------------------------- #
# For Geographic distribution of Customers
# we need to consider states, cities and zipcodes
# statewise customers
state_distribution = master_df["customer_state"].value_counts().head(25)

print(state_distribution)

# graphical representation
plt.figure(figsize=(10,5))

state_distribution.plot(kind="bar")
plt.title("Top 25 Customer States")
plt.xlabel("State")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# -------------------------------------------------------------- #
# ------ Relationship Between Delivery Time And Ratings   ------ #
# -------------------------------------------------------------- #
# co-relate delivery time with ratings
delivery_vs_rating = master_df.groupby("review_score")["delivery_time_days"].mean().reset_index()

print(delivery_vs_rating)

# graphical representation of delivery time in days with review scores
plt.figure(figsize=(8,5))

sns.barplot(data=delivery_vs_rating,x="review_score",y="delivery_time_days")
plt.title("Average Delivery Time by Review Score")
plt.xlabel("Review Score")
plt.ylabel("Average Delivery Time (Days)")
plt.show()

# exact co-relation
correlation = master_df[["review_score","delivery_time_days"]].corr()

print(correlation)
```

```python
# --------------------------------------------------------- #
# ------ Identification Of Dissatisfaction Patterns  ------ #
# --------------------------------------------------------- #

print("-" * 75)
print("ORIGINAL USER-MOVIE RATINGS MATRIX")
print("-" * 75)

# create a list of dissatified customers with rating <=2
dissatisfied = master_df[master_df["review_score"] <= 2]
print(dissatisfied)

# # calculate the delay
master_df["delivery_delay_days"] = (master_df["order_delivered_customer_date"] - master_df["order_estimated_delivery_date"]).dt.days
print(master_df["delivery_delay_days"])
print(dissatisfied["delivery_delay_days"].describe())
print(dissatisfied["order_status"].value_counts())

corr_matrix = master_df[["review_score","delivery_time_days","delivery_delay_days"]].corr()
print(corr_matrix)

plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix,annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

plt.figure(figsize=(10,5))
sns.boxplot(data=master_df,x="review_score",y="delivery_time_days")
plt.title("Delivery Time Distribution by Review Score")
plt.show()
```
